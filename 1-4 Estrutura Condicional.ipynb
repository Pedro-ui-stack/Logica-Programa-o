***Nota do aluno***

```python
nome = input('Digite o nome do aluno(a): ')
nota = float(input('Digite a nota do aluno(a): '))
if nota >= 6.0:
  print('o aluno(a)',nome,'tirou a nota',nota,', está aprovado!')
elif nota > 4.5 and nota < 6.0:
  print('o aluno(a)',nome,'tirou a nota',nota,', o aluno terá que fazer a recuperação.')
else:
  print('o aluno(a)',nome,'tirou a nota',nota,', aluno está reprovador.')
```

**Triângulo**

```python
# é um triângulo ou não
a = int(input('Digite o lado a: '))
b = int(input('Digite o lado b: '))
c = int(input('Digite o lado c: '))
if a + b > c and a + c > b and b + c > a:
  print('É um Triângulo!')
else:
  print('Não é um Triângulo!')
```

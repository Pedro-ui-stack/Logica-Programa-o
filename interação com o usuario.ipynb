Typer

```python
x = 10
type(x)
print('x', type(x))

nome = 'pedro'
print('nome', type(nome))

num = '100'
num_convert = int(num)  # Converter variavel
print('num', type(num))
print('num_convert', type(num_convert))

num_float = float(num)
print('num_float', num_float)

```

Input
* Armazem a informação digitada pelo usuario em uma variavel
* O dado armazenado sempre será um texto

```python
nome = input('Qual é o seu nome: ') # aguarde o usuario digitar algo
print(nome)
```

```python
a = int(input('Digite um numero: '))
b = int(input('Digite outro numero: '))
soma = a + b
print('A soma dos numeros digitados é ', soma)
```

```python
# desconto de um produto
valor_p = int(input('Digite o valor do produto: '))
desconto = valor_p - 10
print('O produto custa',valor_p,', aplicando o desconto de 10 reais, o produto sai no valor de ', desconto)


```

```python
# Outra forma de escrever
valor_p = input('Digite o valor do produto: ')
desconto = int(valor_p) - 10
print('O produto custa',valor_p,', aplicando o desconto de 10 reais, o produto sai no valor de ', desconto)
```

```python
# Solicite o nome do usuario e o exibe
nome_usuario = str(input('Por favor, insira seu nome: '))
print(f'Ola {nome_usuario}!, Seja bem vindo ao sistema.')

# Solicite a idade do usuario e para o inteiro
idade_usuario = int(input('Por favor, insira sua idade: '))

#Verifique se o usuario e maior de idade
if idade_usuario >= 18:
  print('Você é maior de idade.')
else:
  print('Você é menor de idade.')
```
